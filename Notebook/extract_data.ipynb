## Kaggle Datasets

```python
import kagglehub
from pathlib import Path
import shutil

# Define data folder
data_folder = Path.cwd().parent / "Data" / "raw"

# List of datasets to download from Kaggle
list_of_files = [
    # url attrivutes data
    "simaanjali/tes-upload",
    "simaanjali/phising-detection-dataset",
    "nitsey/dataset-phising-website",
    "eswarchandt/phishing-website-detector",
     # phising site urls
    "taruntiwarihp/phishing-site-urls",
     # top searches
    "cheedcheed/top1m"
]

# clean up temp files from previous runs
shutil.rmtree(data_folder)

for dataset in list_of_files:
    # Download latest version
    path = kagglehub.dataset_download(dataset)
    # Move data to project data folder
    # kaggle_dataset_dir = Path(path)
    kaggle_dataset_dir = Path(path).parent.parent.parent.parent

    for item in kaggle_dataset_dir.iterdir():
        print(item)
        dest_path = data_folder / item.name
        print(dest_path)
        shutil.move(str(item), str(dest_path))

```
